#### Groupby :

##### 1. Groupby by Item Type:

```python
# Total Sales By Item Type:
it1=df.groupby('Item Type')['Sales'].sum()
it1
```

```python
# Average Sales by Item Type:
it2=df.groupby('Item Type')['Sales'].mean()
it2
```

```python
# Average Rating by Item Type:
it3=df.groupby('Item Type')['Rating'].mean()
it3
```

##### 2. Groupby by Fat Content:

```python
# Total Sales:
fc1=df.groupby('Item Fat Content')['Sales'].sum()
fc1
```

```python
# Average Sales:
fc2=df.groupby('Item Fat Content')['Sales'].mean()
fc2
```

```python
# Average Rating:
fc3=df.groupby('Item Fat Content')['Rating'].mean()
fc3
```

##### 3. Groupby by Outlet Type:

```python
# Total sales:
ot1=df.groupby('Outlet Type')['Sales'].sum()
ot1
```

```python
# Average Sales:
ot2=df.groupby('Outlet Type')['Sales'].mean()
ot2
```

```python
# Average Rating:
ot3=df.groupby('Outlet Type')['Rating'].mean()
ot3
```

```python
# Average Item Visibility:
ot4=df.groupby('Outlet Type')['Item Visibility'].mean()
ot4
```

##### 4. Multiple Groupby:

```python
# Item Type + Item Fat Content -> Sales:
df.groupby(['Item Type','Item Fat Content'])['Sales'].sum()
```

# Exploratory Data Analysis (EDA) :~

```python
ifc=sns.countplot(x='Item Fat Content',data=df,hue='Outlet Location Type')
plt.title('Item Fat Content')

for bars in ifc.containers:
    ifc.bar_label(bars)
```

```python
it=sns.countplot(x='Item Type',data=df,hue='Outlet Type')
sns.set(rc={'figure.figsize':(18,5)})
plt.title('Item Type')
plt.xticks(rotation=90)

for bars in it.containers:
    it.bar_label(bars)
```

```python
os=sns.countplot(x='Outlet Size',data=df,hue='Outlet Type')
plt.title('Outlet Size')

for bars in os.containers:
    os.bar_label(bars)
```

```python
ot=sns.countplot(x='Outlet Type',data=df,hue='Outlet Size')
plt.title('Outlet Type')

for bars in ot.containers:
    ot.bar_label(bars)
```

```python
# Highest-selling Item Types:
sales_item=df.groupby(['Item Type'],as_index=False)['Sales'].sum().sort_values(by='Sales',ascending=False)
sales_item

sns.barplot(x='Item Type', y='Sales',data=sales_item, hue='Sales')
plt.xticks(rotation=90)
plt.title('Highest-selling Item Types')
plt.show()
```

```python
## Lowest-selling Item Types:
sales_item=df.groupby(['Item Type'],as_index=False)['Sales'].sum().sort_values(by='Sales')
sales_item

sns.barplot(x='Item Type', y='Sales',data=sales_item, hue='Sales')
plt.xticks(rotation=90)
plt.title('Lowest-selling Item Types')
plt.show()
```

```python
# Highest-rated Item Types:
rate_item=df.groupby(['Item Type'],as_index=False)['Rating'].sum().sort_values(by='Rating',ascending=False)
rate_item

sns.barplot(x='Item Type', y='Rating',data=rate_item, hue='Rating')
plt.xticks(rotation=90)
plt.title('Highest-rated Item Types')
plt.show()
```

```python
# Best Outlet Types by Sales:
sales_outlet=df.groupby(['Outlet Type'],as_index=False)['Sales'].sum().sort_values(by='Sales',ascending=False)
sales_outlet

sns.barplot(x='Outlet Type', y='Sales',data=sales_outlet, hue='Sales')
plt.title('Best Outlet Types by Sales')
plt.show()
```

```python
# Top 5 Items by Sales:
items_sales=df.groupby(['Item Type'],as_index=False)['Sales'].sum().sort_values(by='Sales',ascending=False).head()
items_sales

sns.barplot(x='Item Type',y='Sales',data=items_sales,hue='Sales')
sns.set(rc={'figure.figsize':(20,5)})
plt.title('Top 5 Items by Sales')
plt.show()
```

```python
# Bottom 5 Items by Sales:
items_sales=df.groupby(['Item Type'],as_index=False)['Sales'].sum().sort_values(by='Sales').head()
items_sales

sns.barplot(x='Item Type',y='Sales',data=items_sales,hue='Sales')
sns.set(rc={'figure.figsize':(20,5)})
plt.title('Bottom 5 Items by Sales')
plt.show()
```
